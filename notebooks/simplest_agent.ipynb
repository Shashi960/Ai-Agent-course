```python
from agents import Agent, Runner, trace
from openai.types.responses import ResponseTextDeltaEvent
```

Create a simple Nutrition Assistant Agent

```python
nutrition_agent = Agent(
    name="Nutrition Assistant",
    instructions="""
    You are a helpful assistant giving out nutrition advice.
    You give concise answers.
    """,
)
```

Create a Student Management Agent

```python
student_management_agent = Agent(
    name="Student Management Assistant",
    instructions="""
    You are a helpful student management assistant for college students.
    Help with attendance, timetables, exam schedules, fee reminders,
    course registration, deadline tracking, and general student support.
    Keep answers concise, clear, and encouraging.
    """,
)
```

```python
with trace("Student Management Agent"):
    result = await Runner.run(
        student_management_agent,
        "I have a class timetable clash and my attendance is falling. What should I do?",
    )

print(result.final_output)
```

Let's execute the Agent:

```python
with trace("Simple Nutrition Agent"):
    result = await Runner.run(nutrition_agent, "How healthy are bananas?")

print(result)
```

Create a Placement Trainer Agent

```python
placement_trainer = Agent(
    name="Placement Trainer",
    instructions="""
    You are a helpful placement trainer for college students.
    Guide them on resume building, aptitude practice, coding interview prep,
    communication, and mock interviews.
    Keep answers concise, practical, and encouraging.
    Give short step-by-step plans when asked for preparation advice.
    """,
)

```

```python
with trace("Placement Trainer Agent"):
    result = await Runner.run(
        placement_trainer,
        "I have 2 weeks left for placement. What should I do every day?",
    )

print(result.final_output)
```

Streaming the answer to the screen, token by token

```python
response_stream = Runner.run_streamed(nutrition_agent, "How is the placement in bmsit for MCA?")

async for event in response_stream.stream_events():
    if event.type == "raw_response_event" and isinstance(
        event.data, ResponseTextDeltaEvent
    ):
        print(event.data.delta, end="", flush=True)
```
